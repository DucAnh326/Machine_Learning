```python

import numpy as np

def f(x):
    return (1/3) * (x**3) - x

def df(x):
    return x**2 - 1

def GD1(x0, eta=0.1):
    x = [x0]

    for it in range(100):
        if not np.isfinite(x[-1]) or abs(x[-1]) > 1e5:
            return None, it

        grad = df(x[-1])
        if abs(grad) < 1e-4:
            return x, it

        x_new = x[-1] - eta * grad
        x.append(x_new)

    return x, it

(x1, it1) = GD1(5, .1)
print(f"Final x: {x1[-1]:.4f} | Value: {f(x1[-1]):.4f} | Iterations: {it1}" if x1 is not None else "-∞")
```
